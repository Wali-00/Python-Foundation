```python
print("enter 3 number in one line")
inp = input();

container = inp.split()
A = int(container[0])
B = int(container[1])
C = int(container[2])

if A<B and A<C:
    print(A," is min")

elif B<A and B<C:
    print(B, " is min")

else:
    print(C," is min")

if A>B and A>C:
    print(A," is max")

elif (B>A and B>C):
    print(B, " is max")

else:
    print(C," is max")    



```
