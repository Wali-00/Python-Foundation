```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

days = [1,2,3,4,5,6,7]
hours = [2,3,1,4,5,6,2]
plt.plot(days,hours)

#add label 
plt.xlabel('Days')
plt.ylabel('Hours')
#tittle 
plt.title('Daily study hours')
```

```python
days = [1,2,3,4,5,6,7]
hours = [2,3,1,4,5,6,3]
hours1 = [3,1,4,3,4,2,2]
plt.plot(days,hours)
plt.plot(days,hours1)

plt.xlabel('Days')
plt.ylabel('Hours')

plt.title('Daily study hours')
```

```python
df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.plot(df['Year'],df['Programming'])
plt.plot(df['Year'],df['Digital Marketing'])
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')
```

```python
#add legend 
plt.plot(df['Year'],df['Programming'],label='Programming')
plt.plot(df['Year'],df['Digital Marketing'],label= 'Digital Marketing')
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')
plt.legend(loc = 'best')
```
